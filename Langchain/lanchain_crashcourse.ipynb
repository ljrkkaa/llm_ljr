## LLMs

```python
from langchain.llms import OpenAI

from langchain_openai import ChatOpenAI
openai_api_key = "EMPTY"
openai_api_base = "http://127.0.0.1:1234/v1"
llm = ChatOpenAI(
    openai_api_key=openai_api_key,
    openai_api_base=openai_api_base,
    temperature=0.7,
)
# llm = OpenAI(temperature=0.9)
name = llm.predict("I want to open a restaurant for Chinese food. Suggest a fency name for this.")
print(name)
```

```python
response = llm.invoke("I want to open a restaurant for Indian food. Suggest a fancy name for this.")
print(response.content)
```

## Prompt Templates

```python
from langchain.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables =['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this."
)
p = prompt_template_name.format(cuisine="Italian")
print(p)

# print(llm.predict(p))
```

## Chains

```python
# llm = OpenAI(temperature=0.6)

prompt_template_name = PromptTemplate(
    input_variables =['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this."
)

name_chain =LLMChain(llm=llm, prompt=prompt_template_name)

prompt_template_items = PromptTemplate(
    input_variables = ['restaurant_name'],
    template="""Suggest some menu items for {restaurant_name}"""
)

food_items_chain = LLMChain(llm=llm, prompt=prompt_template_items)
```

#### Simple Sequential Chain

```python
from langchain.chains import SimpleSequentialChain
chain = SimpleSequentialChain(chains = [name_chain, food_items_chain])

content = chain.run("Indian")
print(content)
```

#### serpapi and llm-math tool

```python
from langchain.agents import AgentType, initialize_agent, load_tools
from langchain.llms import OpenAI

# llm = OpenAI(temperature=0)

# The tools we'll give the Agent access to. Note that the 'llm-math' tool uses an LLM, so we need to pass that in.
tools = load_tools(["serpapi", "llm-math"], llm=llm)
  
# Finally, let's initialize an agent with the tools, the language model, and the type of agent we want to use.
agent = initialize_agent(tools, llm, agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True)

# Let's test it out!
agent.run("What was the GDP of US in 2022 plus 5?")
```

#### Wikipedia and llm-math tool

```python
# install this package: pip install wikipedia

# The tools we'll give the Agent access to. Note that the 'llm-math' tool uses an LLM, so we need to pass that in.
tools = load_tools(["wikipedia", "llm-math"], llm=llm)

# Finally, let's initialize an agent with the tools, the language model, and the type of agent we want to use.
agent = initialize_agent(
    tools, 
    llm, 
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, 
    verbose=True
)

# Let's test it out!
agent.run("When was Elon musk born? What is his age right now in 2023?")
```

## Memory

```python
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate

prompt_template_name = PromptTemplate(
    input_variables =['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this."
)
chain = LLMChain(llm=llm,prompt=prompt_template_name)
name = chain.run("Mexican")
print(name)
```

```python
name = chain.run("Indian")
print(name)
```

```python
chain.memory
```

```python
type(chain.memory)
```

#### ConversationBufferMemory

```python
from langchain.memory import ConversationBufferMemory

memory = ConversationBufferMemory()

chain = LLMChain(llm=llm, prompt=prompt_template_name, memory=memory)
name = chain.run("Mexican")
print(name)
```

```python
name = chain.run("Arabic")
print(name)
```

```python
print(chain.memory.buffer)
```

#### ConversationChain

```python
convo.run("How much is 5+5?")
```

```python
convo.run("Who was the captain ofthe winning team?")
```

```python
print(convo.memory.buffer)
```

#### ConversationBufferWindowMemory

```python
from langchain.memory import ConversationBufferWindowMemory

# k=1 表示只记住 上一轮对话，也就是窗口大小为 1。
memory = ConversationBufferWindowMemory(k=1) 
 
convo = ConversationChain(
    llm = ChatOpenAI(
    openai_api_key=openai_api_key,
    openai_api_base=openai_api_base,
    temperature=0.7,
),
    memory=memory
)
convo.run("Who won the first cricket world cup?")
```

```python
convo.run("How much is 5+5?")
```

```python
convo.run("Who was the captain of the winning team?")
```

```python
print(convo.memory.buffer)
```
