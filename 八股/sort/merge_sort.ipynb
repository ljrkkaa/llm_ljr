```python
def merge_sort(nums):
    if len(nums) <= 1:
        return nums

    mid = len(nums) // 2
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    return merge(left, right)


def merge(left, right):
    res = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1

    # 收尾
    res.extend(left[i:])
    res.extend(right[j:])

    return res

l = [9,8,7,6,5,4]
print(merge_sort(l))
```

辅助空间优化

```python
class Solution:
    def sortArray(self, nums):
        n = len(nums)
        self.temp = [0] * n
        self.mergeSort(nums, 0, n - 1)
        return nums

    def mergeSort(self, nums, left, right):
        if left >= right:
            return

        mid = left + (right - left) // 2
        self.mergeSort(nums, left, mid)
        self.mergeSort(nums, mid + 1, right)
        self.merge(nums, left, mid, right)

    def merge(self, nums, left, mid, right):
        # 拷贝当前区间
        for i in range(left, right + 1):
            self.temp[i] = nums[i]

        i, j = left, mid + 1
        index = left

        while i <= mid and j <= right:
            if self.temp[i] <= self.temp[j]:   # 注意这里用 <= 保证稳定性
                nums[index] = self.temp[i]
                i += 1
            else:
                nums[index] = self.temp[j]
                j += 1
            index += 1

        while i <= mid:
            nums[index] = self.temp[i]
            i += 1
            index += 1

        while j <= right:
            nums[index] = self.temp[j]
            j += 1
            index += 1
```
