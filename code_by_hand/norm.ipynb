```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import math

class LayerNorm(nn.Module):
    def __init__(self, hidden_dim, eps=1e-6):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.eps = eps
        self.w = nn.Parameter(torch.ones(hidden_dim))
        self.b = nn.Parameter(torch.zeros(hidden_dim))
        
    def forward(self,x):
        mean = x.mean(dim=-1,keepdim=True)
        std = x.std(dim=-1, keepdim=True, unbiased=False)

        x_norm = (x-mean)/(std+self.eps)

        return x_norm * self.w + self.b

x = torch.rand(4,4,128)

ln = LayerNorm(128)

print(ln(x).shape)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F


class RMSNorm(nn.Module):
    def __init__(self, hidden_dim, eps=1e-6):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.eps = eps

        self.gamma = nn.Parameter(torch.ones(hidden_dim))

    def forward(self, x):
        x_norm = x * (torch.rsqrt(x.pow(2).mean(dim=-1,keepdim=True)) + self.eps)

        return x_norm * self.gamma


x = torch.rand(4,4,128)

ln = RMSNorm(128)
x_norm = ln(x)
print(ln(x).shape)

print(torch.norm(x_norm[0, 0, :]))
```
