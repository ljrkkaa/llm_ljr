最小堆的约束条件只有一个：

每个父节点 ≤ 它的所有子节点。

| 操作               | 描述                | 时间复杂度        | 说明                                           |
| ---------------- | ----------------- | ------------ | -------------------------------------------- |
| **heapify**      | 构建堆（把无序数组变成最小堆）   | **O(n)**     | 自底向上调整，复杂度不是 O(n log n)，而是 O(n)（因为下层节点少参与比较） |
| **heappush**     | 插入一个元素            | **O(log n)** | 元素插入最后一个叶节点，然后“上浮”（bubble up）调整堆序            |
| **heappop**      | 删除最小元素（堆顶）        | **O(log n)** | 堆顶与最后一个元素交换，然后“下沉”（bubble down）恢复堆序          |
| **heappushpop**  | 先 push 后 pop（更高效） | **O(log n)** | 比单独执行两步快，因为只需一次调整                            |
| **heapreplace**  | 先 pop 再 push（更高效） | **O(log n)** | 同上                                           |
| **peek (取堆顶元素)** | 直接访问最小值（heap[0]）  | **O(1)**     | 不需要重排                                        |
| **删除任意节点**       | 删除堆中任意位置元素        | **O(log n)** | 删除后要上浮或下沉调整                                  |
| **合并两个堆**        | 把两个堆合并为一个         | **O(n + m)** | 可以先拼接后 heapify                               |


```python
import heapq

lst = [1,2,3,5,1,5,8,9,6]

heapq.heapify(lst)
print(lst)

# 去掉最小的
heapq.heappop(lst)
print(lst)

# 加入100
heapq.heappush(lst,100)
print(lst)

# 去最小 加进99
heapq.heapreplace(lst,99)
print(lst)
```

```python
'''
最大的n个是谁；
最小的n个是谁；
'''
print(heapq.nlargest(3,lst))

print(heapq.nsmallest(3,lst))
```

当列表中的元素是元组时，此时若创建最小堆，则是按照各元组第一个元素的字典序创建的。
