```python
import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
def binary_cross_entropy(y_pred,y_true):
    y_pred = 1/(1+torch.exp(-y_pred))

    epsilon = 1e-8
    y_pred = torch.clamp(y_pred, epsilon, 1. - epsilon)

    bce_loss = - (y_true * torch.log(y_pred) + (1. - y_true) * torch.log(1. - y_pred))

    return torch.mean(bce_loss)

# Example usage
y_pred = torch.tensor([0.2, 0.7, 0.9, 0.4], dtype=torch.float32)
y_true = torch.tensor([0, 1, 1, 0], dtype=torch.float32)
loss = binary_cross_entropy(y_pred, y_true)
print(f'Custom BCE Loss: {loss.item()}')
```

```python
loss = nn.BCEWithLogitsLoss()(y_pred, y_true)
print(f'PyTorch BCEWithLogitsLoss: {loss.item()}')

loss = nn.BCELoss()(torch.sigmoid(y_pred), y_true)
print(f'PyTorch BCELoss with Sigmoid: {loss.item()}')
```
